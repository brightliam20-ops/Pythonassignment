# Question1. How many loans, how much money and what average loan size does each sector have?

```python
sector_summary = df.groupby("sector").agg(
    n_loans=("id", "size"),
    total_lent=("loan_amount", "sum"),
    mean_loan=("loan_amount", "mean")
).sort_values("n_loans", ascending=False)
sector_summary
```

Agriculture has the most loans (180,302) and the most money lent (about 143 million USD), followed by Food and Retail. Sectors with few loans, such as Entertainment and Wholesale, have the largest average loans (about 1,674 and 1,570 USD), while Personal Use has the smallest (411 USD).

# Question2. What are the 10 largest loans above $10,000?

```python
big = df[df["loan_amount"] > 10000]
print("Loans above $10,000:", len(big))
big.sort_values("loan_amount", ascending=False)[
    ["id", "country", "sector", "activity", "loan_amount", "funded_amount", "lender_count"]].head(10)
```

Only 184 loans are above 10,000 USD. The largest is 100,000 USD for agriculture in Haiti, and most of the rest are 50,000 USD loans for renewable energy products or agriculture across Africa and Latin America.

# Question3. Which countries have the most farming-related loans?

```python
farm = df[df["activity"].str.contains("farm", case=False, na=False) |
          df["activity"].str.contains("livestock", case=False, na=False)]
print("Farming-related loans:", len(farm), "of", len(df))
farm.groupby("country").agg(
    n_loans=("id", "size"),
    total_lent=("loan_amount", "sum"),
    mean_loan=("loan_amount", "mean")
).sort_values("n_loans", ascending=False).head(10)
```

90,402 loans (about 13.5% of the data) have "farm" or "livestock" in the activity name. Kenya leads with 21,427 loans, followed by the Philippines (13,923) and Cambodia (8,918).

# Question4. How have the number and total value of loans changed by year?

```python
df["year"] = df["date"].apply(lambda d: d[:4])
df.groupby("year").agg(
    n_loans=("id", "size"),
    total_lent=("loan_amount", "sum"),
    mean_loan=("loan_amount", "mean")
)
```

The number of loans grew from 174,234 in 2014 to a peak of 197,236 in 2016 (164.4 million USD lent). The 2017 row is lower because the data stops on 26 July 2017, and the average loan size fell steadily from about 878 to 767 USD.

# Question5. For education loans in the five most active countries, how do loan size and term compare?

```python
top5 = df["country"].value_counts().head(5).index
edu = df[(df["sector"] == "Education") & (df["country"].isin(top5)) & (df["loan_amount"] >= 100)]
edu.groupby("country").agg(
    n_loans=("id", "size"),
    mean_loan=("loan_amount", "mean"),
    mean_term=("term_in_months", "mean")
).sort_values(["mean_loan", "n_loans"], ascending=[False, False])
```

Education loans are largest in Cambodia (about 1,232 USD) and smallest in the Philippines (about 409 USD). Cambodia's education loans also have by far the longest term, about 40 months compared with 16 to 21 months in the other four countries.

# Question6. Which large loans (at least $5,000) attracted very few lenders (20 or fewer)?

```python
few = df[(df["loan_amount"] >= 5000) & (df["lender_count"] <= 20)]
print("Large loans with few lenders:", len(few), "out of", len(df[df["loan_amount"] >= 5000]))
few.sort_values(["lender_count", "loan_amount"], ascending=[True, False])[
    ["id", "country", "sector", "loan_amount", "funded_amount", "lender_count"]].head(10)
```

2,231 of the 11,683 loans of 5,000 USD or more (19.1%) had 20 or fewer lenders. The top of the list shows loans with 0 lenders and 0 USD funded, including 50,000 USD requests in Mexico and Haiti and many 10,000 USD loans in the United States.
